```
import os, sys
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

def find_project_root():
    curr = os.path.abspath(os.getcwd())
    while curr != os.path.dirname(curr):
        if os.path.exists(os.path.join(curr, 'data', 'processed', 'master_train.csv')):
            return curr
        if os.path.exists(os.path.join(curr, 'team_6', 'data', 'processed', 'master_train.csv')):
            return os.path.join(curr, 'team_6')
        curr = os.path.dirname(curr)
    return os.path.abspath(os.getcwd())

ROOT_DIR = find_project_root()
FIGURES_DIR = os.path.join(ROOT_DIR, 'figures')
os.makedirs(FIGURES_DIR, exist_ok=True)
master_train = pd.read_csv(os.path.join(ROOT_DIR, 'data', 'processed', 'master_train.csv'))
print("Master dataset loaded. Ready to display figures.")
```
